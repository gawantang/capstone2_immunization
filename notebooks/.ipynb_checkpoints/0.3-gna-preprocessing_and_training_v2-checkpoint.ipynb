## Impute remaining vars for X_train with interpolate()

For 2nd batch of features, use a different approach for other features that may have changed over 4 year period
- 'Infant mortality rate'
- % Broadband Access - 5% change between two years of non-null data 
- 'live_births'
- '% Less Than 18 Years of Age'
- 'segregation index_b'
- % Vaccinated Flu
- and more shown below...

```python
vars_to_impute = (['% Low Birthweight',
                    'Infant Mortality Rate',
                    '% Broadband Access', 
                    '# Primary Care Physicians',
                    'live_births',
                    '% Vaccinated Flu',
                    '% Less Than 18 Years of Age', 
                    'Segregation index_b',
                    'other_pc_provider_ratio_numerator',
                    'Primary Care Physicians Rate',
                    'other_pc_provider_ratio_denominator',
                    'Other Primary Care Provider Rate',
                    'kin_HepA_cov',
                    'kin_HepB_cov',
                    'kin_polio_cov']
                 )
X_train[vars_to_impute].columns
```

To use interpolate, which works to replace values in one column at a time, we will need to reshape the data such that a single column or row contains all the values for the four years for each county.

**interpolate**
    # Using 'polynomial' interpolation of order 2
    s = pd.Series([1, 2, np.nan, np.nan, 5, 6])
    s_poly = s.interpolate(method='polynomial', order=2)
    print(s_poly)

```python
print(X_train.loc[:,vars_to_impute].isna().sum())
```

Linear interpolation can fill gaps between existing values. It can't fill gaps when:
- An entire county has all missing values
- All values at the beginning or end of a county's time series are missing

```python
# Apply linear interpolation
#sort rows by county
X_train = X_train.sort_values(by = ['county_name','year'])
# apply interpolate command to one county at a time with year values in different rows
X_train[vars_to_impute] = (
    X_train.groupby('county_name')[vars_to_impute]
        .transform(lambda df: df.interpolate(method='linear'))
        #.reset_index(drop=True)
)
X_train[vars_to_impute].isna().sum()
```

There are reduced numbers of missing values for 
- IMR, broad band, live births, youth population, segregation and primary care physicians rate
- many missing values still remain

Try applying forward and back fill options. 
This still wont fill gaps when:
- An entire county has all missing values
but it will help if...
- All values at the beginning or end of a county's time series are missing

```python
X_train = X_train.sort_values(by=['county_name', 'year'])

# Interpolate, then fill edges
X_train[vars_to_impute] = (
    X_train.groupby('county_name')[vars_to_impute]
        .transform(lambda x: x.interpolate(method='linear').ffill().bfill())
)

X_train[vars_to_impute].isna().sum()
```

Impute remaining values as mean of all other non-null county values. This cant be county specific because many counties are missing all data for a given variable.

```python
# global imputation
for var in vars_to_impute:
    X_train[var].fillna(X_train[var].median(), inplace=True)
X_train[vars_to_impute].isna().sum()
```

```python
# drop original variables with NaN values
X_train = X_train.dropna(axis = 1)
```
