### Create and run a function to extract CHR files

```python
# create a function to extract the CHR files from the other 3 pd.ExcelFile files
# and return the one data set for each Excel sheet 

def extract_chr_columns(excel_file_list, sheet_name_list):
    '''
    Sheet_name_list should specify only two sheet names.

    '''
    #initialize dictionary for saving resulting dataframes
    new_dfs = {}
    
    for i, excel_file in enumerate(excel_file_list):
        print(f"Processing {excel_file}....\n")
        print(f"Sheet names in {excel_file}:")
        # print(excel_file.sheet_names)

        sheet_names_to_extract = sheet_name_list
        # specify the names of the columns to look for in each sheet
        cols_to_select = (['County', 
                           'Infant Mortality Rate',
                           '# Uninsured Children',
                           '% Uninsured Children',
                           '# Uninsured_c',# manually changed column header in excel to avoid pulling adult #'s
                           '% Uninsured_c', # manually changed column header in excel to avoid pulling adult %'s
                           'Other Primary Care Provider Rate',
                           'Other Primary Care Provider Ratio',
                           'Median Household Income',
                           '% Enrolled in Free or Reduced Lunch',
                           '# Food Insecure',
                           '% Food Insecure',
                           'Segregation Index_b', # residential segregation index, black-white
                           'Segregation index_b',
                           'Population',
                           '% Female',
                           '% Rural',
                           '% Broadband Access',
                           '% Less Than 18 Years of Age', 
                           '% Low Birthweight',
                           '% Uninsured',
                           '# Primary Care Physicians',
                           'Primary Care Physicians Rate',
                           '% Vaccinated',
                           '% Children in Poverty'])

        if sheet_names_to_extract[0] in excel_file.sheet_names:
            
            # if sheet is in the ExcelFile, extract data and rename column headers
            first_sheet_df = excel_file.parse(sheet_names_to_extract[0],header=1, skiprows=1)
            first_sheet_df_columns = excel_file.parse(sheet_names_to_extract[0], header=0, skiprows=1).columns
            #print(f"First sheet columns: {first_sheet_df_columns}")
            first_sheet_df.columns = first_sheet_df_columns
            
            # initialize empty list of columns in the sheet
            cols_found = []
            for col in cols_to_select:
                if col in first_sheet_df.columns:
                    cols_found += [col]
            # extract the columns of interest from the sheet
            first_sheet_df= first_sheet_df[cols_found]
            
        else: # if sheet is not in the ExcelFile, continue to next sheet
            print(f"{sheet_names_to_extract[0]} was not found in {excel_file.sheet_names}.")
        
        # if sheet is in the ExcelFile, extract data and rename column headers
        if sheet_names_to_extract[1] in excel_file.sheet_names:
            second_sheet_df = excel_file.parse(sheet_names_to_extract[1],header=1, skiprows=1)
            second_sheet_df_columns = excel_file.parse(sheet_names_to_extract[1], header=0, skiprows=1).columns
            #print(f"Second sheet columns: {second_sheet_df_columns}")
            second_sheet_df.columns = second_sheet_df_columns

            # initialize empty list of columns in the sheet
            cols_found2 = []
            for col in cols_to_select:
                if col in second_sheet_df.columns:
                    cols_found2 += [col]
            # extract the columns of interest from the sheet
            second_sheet_df= second_sheet_df[cols_found2]

        else: # if sheet is not in the ExcelFile, continue on
            print(f"{sheet_names_to_extract[1]} was not found in {excel_file.sheet_names}.")
        
        print(f"Saving data frame for {excel_file}....\n")

        # combine columns from two different sheets of the data frames for that year
        new_df = pd.merge(first_sheet_df, second_sheet_df, on="County", how="outer", validate="one_to_one")
        
        print(new_df.shape)
        print(new_df.columns)
        # excel_file + '_df' = new_df

        # save resulting data frames in dictionary
        # new_dfs[excel_file] = new_df
        new_dfs[f"df_{i}"] = new_df

    return new_dfs
```

- **Social Vulnerability Index (SVI)** is a Centers for Disease Control and Prevention (CDC)‘s index. It includes socioeconomic, population, and housing/transportation vulnerability measures with scores ranging from 0 representing lowest vulnerability to 1 representing highest vulnerability [link](https://www.atsdr.cdc.gov/placeandhealth/svi/index.html). These counties have also been designated as health professional shortage areas.

- **The COVID-19 Vaccine Coverage Index (CVAC)** The CVAC captures supply- and demand-related challenges that may hinder rapid, widespread COVID-19 vaccine coverage in U.S. counties, through five specific themes: historic undervaccination, sociodemographic barriers, resource-constrained healthcare system, healthcare accessibility barriers, and irregular care-seeking behaviors. The CVAC measures the level of concern for a difficult rollout on a range from 0 (lowest concern) to 1 (highest concern). The CVAC Index can also be categorized as follows: Very Low (0.0-0.19), Low (0.20-0.39), Moderate (0.40-0.59), High (0.60-0.79), or Very High (0.80-1.0) Concern.

- It is calculated based on factors such as access to healthcare, affordable housing, transportation, childcare, and secure employment.

[CDC source](https://data.cdc.gov/stories/s/Vaccine-Hesitancy-for-COVID-19/cnd2-a6zw/). A higher CVAC score indicates greater vulnerability within a community to handle a COVID-19 outbreak.

(Surgo's CVAC Index)[https://vaccine.precisionforcovid.org/ ]
